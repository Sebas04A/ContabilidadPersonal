```python


txt_files_names = ["completo2026_01_14.xlsx", "completo_2025-09-20.xlsx", "diferencias_2025-09-20.xlsx", "diferencias_atipicas3.xlsx"]
lista_dfs = [] # Es mejor guardar todo en una lista y concatenar al final

for file_name in txt_files_names:
    path = os.path.join(project_root, "data", "descripciones", file_name)
    
    # IMPORTANTE: Usamos read_excel porque la extensión es .xlsx
    try:
        df = pd.read_excel(path)
        print(f"Leído exitosamente: {file_name}")
        lista_dfs.append(df)
    except Exception as e:
        print(f"Error al leer {file_name}: {e}")

# Concatenamos todos los DataFrames de la lista
if lista_dfs:
    df_unido = pd.concat(lista_dfs, ignore_index=True)
    
    # Asegúrate de que la columna FECHA sea de tipo datetime antes de ordenar
    df_unido['FECHA'] = pd.to_datetime(df_unido['FECHA'], errors='coerce')
    df_unido.sort_values(by="FECHA", inplace=True)
    
    print("\nDataFrame unido y ordenado por fecha:")
    print(df_unido.head())
else:
    print("No se cargó ningún dato.")
```
