# Debugging Dashboard Transforms
Este notebook te permite probar `transform_metrics` inyectándole manualmente los datos necesarios (ya que `get_processed_data` normal no tiene todas las columnas juntas).

```python
import sys
import os
import pandas as pd

# Asegurar que podemos importar contabilidad
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), '..')))

from contabilidad.backend.storage.data_pipeline import get_pipeline
from contabilidad.config import Col
from contabilidad.backend.storage.transformations.dashboard_transforms import transform_virtual_items, transform_metrics

pipeline = get_pipeline()

# OPCIÓN 1: Usar el daily (que sí tiene todas las columnas unidas) pero solo para ver la tabla cruda
df_crudo_unido = pipeline.get_daily_data('all')
print("Columnas antes de transformaciones:", df_crudo_unido.columns.tolist())
df_crudo_unido.head(3)
```

```python
# Ahora ejecutamos SÓLO los items virtuales
df_virtual = transform_virtual_items(df_crudo_unido)
print("Columnas tras items virtuales:", df_virtual.columns.tolist())
df_virtual[[Col.FECHA, Col.PAGOS_FIJOS, Col.INTERPOLADO]].tail(3)
```

```python
# Y finalmente corremos LAS MÉTRICAS para debuggear tus columnas DIFF
df_final = transform_metrics(df_virtual)
print("Columnas finales:", df_final.columns.tolist())

cols_a_ver = [
    Col.FECHA, Col.TOTAL,
    Col.DIFF_TOTAL, Col.DIFF_INTERPOLADOS, Col.DIFF_TARJETA, Col.DIFF_PAGOS_FIJOS, Col.DIFF_DEUDA_ACUMULADA
]

# Vemos los últimos 10 días para ver los diffs en acción
df_final[cols_a_ver].tail(10)
```
